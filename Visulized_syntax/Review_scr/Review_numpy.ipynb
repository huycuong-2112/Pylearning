### Những cách tạo một mảng bởi numpy

#### Create list with copy() and [:n]

```python
#init 
a = [1,2,3,4]
a = np.array (a)
#work as pointer
b = a[:2]
b+=1
#create new list

c = a[:2].copy()
c+=1
```

```python

print ("a =", a , "b= ", b)
b -=1  #return early a_list
print ("a =", a, "c =", c)
```

#### Create list with range () and linspace()

```python
a = np.arange(0,6,1) # begin,end-1,step
print ("a =", a)

a = np.linspace(0,6,3) # begin,end-1,num_elements
print ("a =", a)
```

#### Create list and join with block()

```python
A = np.zeros((2,2))
B = np.ones((2,2))
C = np.diag([2,3])
D = np.eye(2)

BLOCK = np.block([[A,B],[C,D]])

# print (A)
# print (B)
# print (C)
# print (D)

print (BLOCK)
```

Z-Score formula: 
$$Z = \frac{X-\mathrm{mean}(X)}{\mathrm{sd}(X)}$$

```python
def Zscore(x):
    return (x - np.mean(x))/np.std(x)
```

```python
print (B)
B = Zscore (B)
print (B)
```

Min-Max formula: 
$$Y = \frac{Z-\min(Z)}{\max(Z)-\min(Z)}$$


```python
def Min_Max(x):
    return (x - x.min(axis = None))/ (x.max(axis = None) - x.min(axis = None))
```

```python
B = np.arange(35).reshape(5,7)
B = Min_Max(B)
print (B)
```
