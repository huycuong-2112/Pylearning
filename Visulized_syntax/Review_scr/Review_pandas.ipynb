### RESHAPE THE TABBLE WITH SORT_VALUE & SORT_INDEX and PIVOT, MELT


```python
air_conditions = pd.read_csv("./Dataset/air_quality/air_quality_long.csv", index_col= "date.utc", parse_dates= True).rename(
    columns= str.title
)
air_conditions.head()
air_conditions["Parameter"].value_counts()
no2 = air_conditions[air_conditions["Parameter"] == "no2"]
#no2 = no2.sort_index().groupby(["city"])
no2.sort_values(by=(["Value"]), ascending= False).head()
```

#### LONG TO WIDE BY PIVOT

```python
no2_pivot = no2.pivot(columns="Location", values= "Value")
no2_pivot.head()
#no2_pivot.plot.area(figsize = (12,4), subplots= True)
```

```python
air_pivot_table = air_conditions.pivot_table(columns="Parameter", index ="Location", values= "Value", aggfunc = "sum", margins = True)
#titanic.pivot_table(columns = "Survived", index = "Pclass", values = "Age", aggfunc = "count")
air_pivot_table.head()
```

#### WIDE to LONG BY MELT

```python
no2 = no2_pivot.reset_index()

no2 = no2.melt(id_vars= "date.utc",var_name = "City", value_vars= ["BETR801", "FR04014", "London Westminster"], value_name= "N0_2")
no2.head()
```
