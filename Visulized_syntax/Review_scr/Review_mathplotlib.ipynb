```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
air_quality = pd.read_csv("./Dataset/air_quality/air_quality_no2.csv",index_col= 0,  parse_dates= True)
```

```python
#air_quality.plot()
air_quality.plot.scatter(x = "station_paris", y = "station_london" ,alpha = 0.2)
air_quality.plot.box(figsize=(24, 4), subplots=True )
air_quality.plot.area(figsize=(12, 4), subplots=True)
```

```python
air_quality["station_antwerp"].plot(figsize=(24,6))
plt.show()
air_quality["station_london"].plot(figsize=(24,4))
```

```python
print(air_quality.shape)
city_name = "station_paris"
station_paris= air_quality[air_quality[city_name].notna()]
city = station_paris[city_name]
print (city)
print (city.describe())
print (city.shape)
city.plot.area(figsize=(24,4))
```

```python
fig, axs = plt.subplots(figsize = (12,4))
city.plot.area(ax = axs)
axs.set_ylabel = "No$2_concentrate_city"
axs.set_xlabel = "City"
fig.savefig("Paris_air_quality.png")
```
